# A8) Arithmetic in Python

Computer programming languages are often designed to handle arithmetic
operations such as addition, subtraction, multiplication, division, and
exponents. These operations are performed through the use of operators,
several of which are listed below:

$$+, -, *, /, //, \%, **$$

Just as in mathematics, we can use these operators to develop complex
mathematical equations. Also like in math in a more general sense,
parentheses can be used to give some operations a higher priority than
they would normally have. For example:

$$2 + 4 * 2 = 2 + 8 = 10$$ $$(2 + 4) * 2 = 6 * 2 = 12$$
