-   Will the following recursive function always halt? Why or why not?

````python
def func(num, adjust=0):
    if num % 13 ==0:
        return adjust
    else:
        return func(num+1, adjust+1)
````

-   Will the following recursive function always halt? Why or why not?

````python
def func(num):
    if num <= 1:
        return num
    else:
        return num + func(num+1)
````

-   Below is an example of a recursive function to calculate Fibonacci
    numbers.
    -   What do you think the pros and cons of this method are?

````python
def fib(num):
    if num <= 1:
        return num
    else:
        return fib(num-1) + fib(num-2)
````

-   Below is another example of a recursive function to calculate
    Fibonacci numbers.
    -   Is this method better or worse than the last one? Why?

````python
def fib(num, val=1, prev=0):
    if num == 0:
        return num
    return fib(num-1, val+prev, val)
````
