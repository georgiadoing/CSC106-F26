# Leap years

A year is a leap year if it is divisible by 4, unless it is a century year (divisible by 100) that is not divisible by 400.
Some examples:


* 2020 was a leap year because it is divisible by 4.
* 1900 was not a leap year, because it is divisible by 100, but not 400
* 2000 was a leap year, because it is divisible by 400



```python
# Write a function that takes a year as a parameter and determines whether or not the given year is a leap year.

def is_leap_year(year):
    """Determine if the given year is a leap year."""
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(year," is a leap year.")
    else:
        print(year," is not a leap year.")
```

# Rolling Dice

1.	Write a function that takes two parameters, faces and num_dice that indicate how many sides a single die has and how many dice it should roll respectively.
    +	random.randint may be helpful here.
2.	Add two while loops that check the numbers the user enters:
    + 	The first loop should check the number of faces and should continue running until the user provides an integer larger than one
    + 	The second loop should check the number of dice the user will roll and should continue running until the user enters an integer larger than zero
    + 	These loops ensure that we have at least one side on each “die” and that we roll at least one die.
3.	Call the function you wrote to roll the dice
4.	Add another while loop that encapsulates the previous two loops as well as the call to the function.
    +	This loop should ask the user if they want to roll another set of dice (y/n) and continue so long as the user indicates they want to coninue


```python
# rolling dice
# example input: 6 1

import random
def roll_dice(faces, num_dice):
    """Roll dice and return the results."""
    for i in range(num_dice):
        roll_result=random.randint(1, faces)
    return roll_result

# Loop to ensure valid number of faces
faces = 0
while faces <= 1:
        faces = int(input("Enter the no. of faces on the die (greater than 1): "))
# Loop to ensure valid number of dice
num_dice = 0
while num_dice <= 0:
        num_dice = int(input("Enter the no. of dice to roll (greater than 0): "))
```

```python
# rolling dice - continuous
# example input: y  n
continue_rolling = input("Do you want to roll a set of dice? (y/n): ")
while continue_rolling == 'y':
    # Loop to ensure valid number of faces
    faces = 0
    while faces <= 1:
        faces = int(input("Enter the no. of faces on the die (greater than 1): "))
    # Loop to ensure valid number of dice
    num_dice = 0
    while num_dice <= 0:
        num_dice = int(input("Enter the no. of dice to roll (greater than 0): "))

    # Roll the dice and display results
    results = roll_dice(faces, num_dice)
    print(f"You rolled: {results}")
    # Ask if the user wants to roll again
    continue_rolling = input("Do you want to roll another set of dice? (y/n): ")
```

# Running for Congress

A person is eligible to be a US senator if they are at least 30 years old and have been a US citizen for at least 9 years. To be a US representative, you need to be at least 25 years old and have been a US citizen for at least 7 years. Write a function called eligible_to_run that takes two numbers indicating a person’s age and years of citizenship, then prints a message stating their eligibility for the Senate and House.

The message should have one of the following forms:

*	Congratulations!
You are eligible to run for the House of Representatives and the Senate.
*	You are eligible to run for the House of Representatives, but not for the Senate.
*	Sorry! You are not eligible to run for Congress.

Make sure to test your implementation thoroughly. What are the different scenarios that should be tested? Show what you test by including a list of all function calls at the bottom of your file.

### (Optional 1)

In addition to the citizenship and age requirements, an individual must be a resident of the state they are seeking to represent. Adjust the function you wrote to consider whether the person is a resident of the state or not.


```python
# running for congress

def eligible_to_run(age, years_citizenship, is_resident=False):
    """Check eligibility for US Senate and House of Representatives."""
    if is_resident:
        #print("You are a resident of the state.")
        if  age >= 30 and years_citizenship >= 9:
            print("Congratulations! You are eligible to run for the House of Representatives and the Senate.")
        elif age >= 25 and years_citizenship >= 7:
            print("You are eligible to run for the House of Representatives, but not for the Senate.")
        else:
            print("Sorry! You are not eligible to run for Congress.")
    else:
        print("Sorry! You are not eligible to run for Congress.")
```

```python
eligible_to_run(8,7,True)
eligible_to_run(100,7,True)
eligible_to_run(26,8,True)
eligible_to_run(50,50,False)
eligible_to_run(50,50,True)
```
